```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('sleep_dataset.csv')
df
```

```python
df.isnull().sum().sum()      #for checking null total values
```

```python
df.isnull().sum()     # for checking how much zeros in each column
```

```python
df.info()
```

```python
# axis = 1 for column drop
df = df.drop('Person ID' , axis = 1)
df
```

```python
df.info()
```

```python
for i in df.columns:
    if(df[i].dtype == 'object'):
        print(i , df[i].value_counts())
```

```python
df.shape
```

# Label Encoding

```python
from sklearn.preprocessing import LabelEncoder
Tan = LabelEncoder()
```

```python
for T in df.columns:
    if(df[T].dtype == 'object'):
        df[T] = Tan.fit_transform(df[T])
        print(T , df[T].value_counts())
```

```python
df.info()           #all object datatype is convert into int datatype
```

```python
df.isna().sum().sum()
```

```python
df.shape
```

# Correlation

```python
correlation = df.corr()
correlation
```

```python
import matplotlib.pyplot as plt             #for heat map library include
import seaborn as sns


# constructing a heatmap to understand the correlation

plt.figure(figsize=(20,20))
sns.heatmap(correlation, square=True, fmt='.1f', annot=True, annot_kws={'size':12})
```

# Data Spliting

```python
X=df.drop("Sleep Disorder", axis = 1)                 #data is split on the base of sleep disorder
Y = df["Sleep Disorder"]
X.info()
X.shape
```

```python
Y.info()
Y.shape
```

```python
from sklearn.model_selection import train_test_split
X_TRAIN , X_TEST , Y_TRAIN, Y_TEST = train_test_split(X,Y, test_size =0.30, random_state=200)
print("Size of Train X = " , len(X_TRAIN))
print("Size of Train Y = " , len(Y_TRAIN))
print("Size of Test X = " , len(X_TEST))
print("Size of Test Y = " , len(Y_TEST))
```

# GaussianNB Base Classifier

```python
from sklearn.naive_bayes import GaussianNB
Tanzila = GaussianNB()

Tanzila.fit(X_TRAIN, Y_TRAIN)
GaussianNB()

training_data_prediction = Tanzila.predict(X_TRAIN)                # accuracy for prediction on training data
print(training_data_prediction)
```

# Prediction on Train Data

```python
from sklearn.metrics import accuracy_score
accuracy = accuracy_score(Y_TRAIN, training_data_prediction)
print("The Accuracy of Prediction is ", accuracy)
```

```python
from sklearn.metrics import classification_report, confusion_matrix
confusion_matrix(Y_TRAIN, training_data_prediction)
```

```python
import matplotlib.pyplot as plt
tan = confusion_matrix(Y_TRAIN, training_data_prediction)
fig, ax = plt.subplots(figsize=(7,7)) 
ax.imshow(tan)
ax.xaxis.set(ticks=(0,1,2), ticklabels=('Predicted 0s','Predicted 1s','Predicted 2s'))
ax.yaxis.set(ticks=(0,1,2), ticklabels=('Actual 0s','Actual 1s','Actual 2s'))
for i in range(3):
    for j in range(3):
        ax.text(j, i, tan[i, j], ha='center', va='center', color='white')
        plt.show()
```

```python
report = classification_report(Y_TRAIN, training_data_prediction)
print(report)
```

# GridSearchCV

```python
from sklearn.model_selection import GridSearchCV
# Hyperparameter to fine tune
param_grid = {
 'max_depth': range(1, 10, 1),
 'min_samples_leaf': range(1, 20, 2),
 'min_samples_split': range(3, 20, 2),
 'criterion': ["entropy", "gini"]
}
# Decision tree classifier
tree = DecisionTreeClassifier(random_state=200)
# GridSearchCV
# CV=Cross Validation
grid_search = GridSearchCV(estimator=tree, param_grid=param_grid, cv=5, verbose=True)
grid_search.fit(X_TRAIN, Y_TRAIN)
```

```python
# Best score and estimator
print("best accuracy", grid_search.best_score_)
print(grid_search.best_estimator_)
```
